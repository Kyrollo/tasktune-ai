````python
# =============================================================================
# SQL layer (views-only agent + RBAC)
# =============================================================================


def init_db() -> None:
    """Create schema, views, and seed sample data for TaskTune demo DB."""
    conn = sqlite3.connect(DB_PATH)
    c = conn.cursor()

    c.executescript(
        """
    CREATE TABLE IF NOT EXISTS Companies (
        CompanyId   INTEGER PRIMARY KEY AUTOINCREMENT,
        CompanyName TEXT NOT NULL
    );
    CREATE TABLE IF NOT EXISTS Departments (
        DepartmentId   INTEGER PRIMARY KEY AUTOINCREMENT,
        CompanyId      INTEGER,
        DepartmentName TEXT NOT NULL
    );
    CREATE TABLE IF NOT EXISTS Positions (
        PositionId     INTEGER PRIMARY KEY AUTOINCREMENT,
        CompanyId      INTEGER,
        DepartmentId   INTEGER,
        PositionTitle  TEXT NOT NULL,
        HierarchyLevel INTEGER DEFAULT 1
    );
    CREATE TABLE IF NOT EXISTS Users (
        UserId       INTEGER PRIMARY KEY AUTOINCREMENT,
        CompanyId    INTEGER,
        DepartmentId INTEGER,
        PositionId   INTEGER,
        Email        TEXT NOT NULL,
        FullName     TEXT NOT NULL,
        Role         INTEGER DEFAULT 1
    );
    CREATE TABLE IF NOT EXISTS Projects (
        ProjectId   INTEGER PRIMARY KEY AUTOINCREMENT,
        CompanyId   INTEGER,
        ProjectName TEXT NOT NULL,
        Description TEXT,
        ManagerId   INTEGER,
        Status      INTEGER DEFAULT 1,
        StartDate   TEXT,
        EndDate     TEXT
    );
    CREATE TABLE IF NOT EXISTS Tasks (
        TaskId      INTEGER PRIMARY KEY AUTOINCREMENT,
        ProjectId   INTEGER,
        TaskName    TEXT NOT NULL,
        Description TEXT,
        AssignedTo  INTEGER,
        Priority    INTEGER DEFAULT 2,
        Status      INTEGER DEFAULT 1,
        Deadline    TEXT
    );
    CREATE TABLE IF NOT EXISTS Subtasks (
        SubtaskId   INTEGER PRIMARY KEY AUTOINCREMENT,
        TaskId      INTEGER,
        SubtaskName TEXT NOT NULL,
        Status      INTEGER DEFAULT 1,
        OrderIndex  INTEGER DEFAULT 0
    );
    CREATE TABLE IF NOT EXISTS TaskProgress (
        ProgressId           INTEGER PRIMARY KEY AUTOINCREMENT,
        TaskId               INTEGER,
        SubmittedBy          INTEGER,
        CompletionPercentage INTEGER DEFAULT 0,
        Notes                TEXT
    );
    CREATE TABLE IF NOT EXISTS Evaluations (
        EvaluationId         INTEGER PRIMARY KEY AUTOINCREMENT,
        TaskId               INTEGER,
        EvaluatedUserId      INTEGER,
        EvaluatorId          INTEGER,
        OverallScore         REAL,
        QualityScore         REAL,
        TimelinessScore      REAL,
        CollaborationScore   REAL,
        RiskLevel            INTEGER DEFAULT 1,
        ManualFeedback       TEXT
    );
    CREATE TABLE IF NOT EXISTS Notifications (
        NotificationId   INTEGER PRIMARY KEY AUTOINCREMENT,
        UserId           INTEGER,
        NotificationType INTEGER DEFAULT 1,
        Title            TEXT NOT NULL,
        Message          TEXT,
        RelatedTaskId    INTEGER,
        IsRead           INTEGER DEFAULT 0,
        SentAt           TEXT DEFAULT CURRENT_TIMESTAMP
    );
    CREATE TABLE IF NOT EXISTS AuditLogs (
        LogId        INTEGER PRIMARY KEY AUTOINCREMENT,
        UserId       INTEGER,
        Action       TEXT NOT NULL,
        ResourceType TEXT,
        ResourceId   INTEGER,
        ChangesJson  TEXT,
        Timestamp    TEXT DEFAULT CURRENT_TIMESTAMP
    );
    CREATE TABLE IF NOT EXISTS Reports (
        ReportId          INTEGER PRIMARY KEY AUTOINCREMENT,
        CompanyId         INTEGER,
        GeneratedBy       INTEGER,
        ReportType        INTEGER DEFAULT 1,
        ReportPeriodStart TEXT,
        ReportPeriodEnd   TEXT,
        MetadataJson      TEXT,
        GeneratedAt       TEXT DEFAULT CURRENT_TIMESTAMP
    );
    """
    )

    c.executescript(
        """
    DROP VIEW IF EXISTS TaskFull;
    CREATE VIEW TaskFull AS
    SELECT
        t.TaskId,
        t.TaskName,
        t.Description    AS TaskDescription,
        u.FullName       AS AssignedTo,
        u.Email          AS AssigneeEmail,
        d.DepartmentName AS Department,
        p.ProjectName,
        CASE t.Priority
            WHEN 1 THEN 'Low'
            WHEN 2 THEN 'Medium'
            WHEN 3 THEN 'High'
            ELSE 'Unknown' END AS Priority,
        CASE t.Status
            WHEN 1 THEN 'ToDo'
            WHEN 2 THEN 'InProgress'
            WHEN 3 THEN 'Blocked'
            WHEN 4 THEN 'Done'
            ELSE 'Unknown' END AS Status,
        t.Deadline,
        CASE
            WHEN t.Status != 4 AND t.Deadline < date('now') THEN 'Overdue'
            WHEN t.Status != 4 AND t.Deadline <= date('now', '+7 days') THEN 'DueSoon'
            ELSE 'OnTrack'
        END AS DeadlineStatus,
        tp.CompletionPercentage,
        tp.Notes         AS ProgressNotes
    FROM Tasks t
    LEFT JOIN Users       u  ON t.AssignedTo  = u.UserId
    LEFT JOIN Departments d  ON u.DepartmentId = d.DepartmentId
    LEFT JOIN Projects    p  ON t.ProjectId    = p.ProjectId
    LEFT JOIN (
        SELECT TaskId, CompletionPercentage, Notes
        FROM TaskProgress
        WHERE ProgressId IN (
            SELECT MAX(ProgressId) FROM TaskProgress GROUP BY TaskId
        )
    ) tp ON t.TaskId = tp.TaskId;

    DROP VIEW IF EXISTS EvaluationView;
    CREATE VIEW EvaluationView AS
    SELECT
        e.EvaluationId,
        eu.FullName      AS EvaluatedUser,
        ev.FullName      AS EvaluatedBy,
        t.TaskName,
        e.OverallScore,
        e.QualityScore,
        e.TimelinessScore,
        e.CollaborationScore,
        CASE e.RiskLevel
            WHEN 1 THEN 'Low'
            WHEN 2 THEN 'Medium'
            WHEN 3 THEN 'High'
            ELSE 'Unknown' END AS RiskLevel,
        e.ManualFeedback
    FROM Evaluations e
    LEFT JOIN Users eu ON e.EvaluatedUserId = eu.UserId
    LEFT JOIN Users ev ON e.EvaluatorId     = ev.UserId
    LEFT JOIN Tasks t  ON e.TaskId          = t.TaskId;

    DROP VIEW IF EXISTS NotificationsView;
    CREATE VIEW NotificationsView AS
    SELECT
        n.NotificationId,
        u.FullName  AS RecipientName,
        n.Title,
        n.Message,
        CASE n.NotificationType
            WHEN 1 THEN 'Task'
            WHEN 2 THEN 'Deadline'
            WHEN 3 THEN 'Evaluation'
            WHEN 4 THEN 'System'
            ELSE 'Other' END AS NotificationType,
        t.TaskName  AS RelatedTask,
        CASE n.IsRead
            WHEN 0 THEN 'Unread'
            ELSE 'Read' END AS ReadStatus,
        n.SentAt
    FROM Notifications n
    LEFT JOIN Users u ON n.UserId        = u.UserId
    LEFT JOIN Tasks t ON n.RelatedTaskId = t.TaskId;
    """
    )

    today = datetime.now()
    c.execute(
        "INSERT OR IGNORE INTO Companies(CompanyId,CompanyName) VALUES(1,'TechCorp Egypt')"
    )

    for dept in [(1, 1, 'Engineering'), (2, 1, 'HR'), (3, 1, 'Product'), (4, 1, 'Finance')]:
        c.execute('INSERT OR IGNORE INTO Departments VALUES(?,?,?)', dept)

    for pos in [
        (1, 1, 1, 'Senior Engineer', 3),
        (2, 1, 1, 'Junior Engineer', 1),
        (3, 1, 2, 'HR Manager', 3),
        (4, 1, 3, 'Product Manager', 3),
        (5, 1, 1, 'Tech Lead', 4),
    ]:
        c.execute('INSERT OR IGNORE INTO Positions VALUES(?,?,?,?,?)', pos)

    for u in [
        (1, 1, 1, 5, 'ahmed@tech.com', 'Ahmed Hassan', 3),
        (2, 1, 1, 2, 'sara@tech.com',  'Sara Mohamed', 1),
        (3, 1, 1, 2, 'omar@tech.com',  'Omar Ali',     1),
        (4, 1, 2, 3, 'mona@tech.com',  'Mona Khaled',  2),
        (5, 1, 3, 4, 'tarek@tech.com', 'Tarek Nasser', 2),
        (6, 1, 1, 1, 'layla@tech.com', 'Layla Sami',   1),
    ]:
        c.execute('INSERT OR IGNORE INTO Users VALUES(?,?,?,?,?,?,?)', u)

    for p in [
        (1, 1, 'AI Assistant Platform', 'Build internal AI tools', 1, 2, '2025-01-01', '2025-12-31'),
        (2, 1, 'HR Portal Upgrade',     'Upgrade HR system',       4, 1, '2025-02-01', '2025-09-30'),
        (3, 1, 'Mobile App v2',         'New mobile features',     5, 2, '2025-03-01', '2025-10-31'),
    ]:
        c.execute('INSERT OR IGNORE INTO Projects VALUES(?,?,?,?,?,?,?,?)', p)

    tasks = [
        (1,1,'Build RAG Module',   'Implement vector search', 2,3,2,(today+timedelta(days=7)).strftime('%Y-%m-%d')),
        (2,1,'Design DB Schema',   'Create database design',  3,2,4,(today-timedelta(days=5)).strftime('%Y-%m-%d')),
        (3,1,'API Integration',    'Connect external APIs',   2,2,3,(today-timedelta(days=2)).strftime('%Y-%m-%d')),
        (4,2,'Update HR Policies', 'Review and update',       4,1,2,(today+timedelta(days=14)).strftime('%Y-%m-%d')),
        (5,3,'UI Redesign',        'New mobile UI',           5,2,2,(today+timedelta(days=5)).strftime('%Y-%m-%d')),
        (6,1,'Testing & QA',       'Full system testing',     3,2,1,(today+timedelta(days=10)).strftime('%Y-%m-%d')),
        (7,2,'Employee Onboarding','Setup new hire flow',     4,2,2,(today-timedelta(days=1)).strftime('%Y-%m-%d')),
        (8,3,'Push Notifications', 'Mobile notifications',    6,3,2,(today+timedelta(days=3)).strftime('%Y-%m-%d')),
        (9,1,'Security Audit',     'System security check',   1,3,1,(today+timedelta(days=20)).strftime('%Y-%m-%d')),
    ]
    for t in tasks:
        c.execute('INSERT OR IGNORE INTO Tasks VALUES(?,?,?,?,?,?,?,?)', t)

    for p in [
        (1,1,2,75,  'RAG module 75% complete â€” vector search working'),
        (2,2,3,100, 'Schema finalized and reviewed'),
        (3,3,2,30,  'API integration started, auth done'),
        (4,4,4,50,  'Half of policies updated'),
        (5,5,5,60,  'UI mockups approved, coding started'),
        (6,6,3,10,  'QA plan created'),
        (7,7,4,80,  'Onboarding flow 80% done'),
        (8,8,6,45,  'Notification service in progress'),
    ]:
        c.execute(
            'INSERT OR IGNORE INTO TaskProgress(ProgressId,TaskId,SubmittedBy,CompletionPercentage,Notes) VALUES(?,?,?,?,?)',
            p,
        )

    for e in [
        (1,2,3,1,90,88,95,87,1,'Excellent schema work, delivered on time'),
        (2,7,4,1,72,70,65,80,2,'Good effort but timeline was missed'),
        (3,5,5,1,85,90,80,85,1,'Great UI work, good collaboration'),
        (4,3,2,1,60,65,55,60,3,'API integration delayed, needs improvement'),
    ]:
        c.execute(
            'INSERT OR IGNORE INTO Evaluations(EvaluationId,TaskId,EvaluatedUserId,EvaluatorId,OverallScore,QualityScore,TimelinessScore,CollaborationScore,RiskLevel,ManualFeedback) VALUES(?,?,?,?,?,?,?,?,?,?)',
            e,
        )

    for n in [
        (1,2,2,'Task Overdue',       'Your task API Integration is overdue',       3,0),
        (2,4,1,'Deadline Reminder',  'HR Policies task due in 14 days',            4,0),
        (3,2,3,'New Evaluation',     'You have received a new evaluation score: 90/100', None,1),
        (4,3,2,'Task Blocked',       'API Integration is blocked',                 3,0),
        (5,5,3,'Task DueSoon',       'UI Redesign due in 5 days',                  5,0),
    ]:
        c.execute(
            'INSERT OR IGNORE INTO Notifications(NotificationId,UserId,NotificationType,Title,Message,RelatedTaskId,IsRead) VALUES(?,?,?,?,?,?,?)',
            n,
        )

    conn.commit()
    conn.close()


def run_sql(query: str):
    """Execute SQL and return rows as list of dicts."""
    try:
        conn = sqlite3.connect(DB_PATH)
        conn.row_factory = sqlite3.Row
        c = conn.cursor()
        c.execute(query)
        rows = [dict(r) for r in c.fetchall()]
        conn.close()
        return rows, None
    except Exception as e:
        return [], str(e)


VIEWS_SCHEMA = """
You have access to ONLY these 3 views â€” do NOT query raw tables directly:

1. TaskFull â€” All tasks with full context:
   Columns: TaskId, TaskName, TaskDescription, AssignedTo (full name),
            AssigneeEmail, Department, ProjectName, Priority (Low/Medium/High),
            Status (ToDo/InProgress/Blocked/Done), Deadline (YYYY-MM-DD),
            DeadlineStatus (Overdue/DueSoon/OnTrack),
            CompletionPercentage (0-100), ProgressNotes

2. EvaluationView â€” Evaluation scores per user/task:
   Columns: EvaluationId, EvaluatedUser (full name), EvaluatedBy (full name),
            TaskName, OverallScore, QualityScore, TimelinessScore,
            CollaborationScore, RiskLevel (Low/Medium/High), ManualFeedback

3. NotificationsView â€” Notifications per user:
   Columns: NotificationId, RecipientName (full name), Title, Message,
            NotificationType (Task/Deadline/Evaluation/System/Other),
            RelatedTask (task name), ReadStatus (Read/Unread), SentAt
"""


def sql_agent(question: str, user_role: int = 1, user_name: str | None = None) -> tuple:
    """Translate natural language to SQL on views (RBAC-aware). Returns (rows, sql, err)."""
    client = _get_groq()
    role_label = {1: 'Employee', 2: 'Manager', 3: 'Admin'}.get(user_role, 'Employee')

    if user_role == 1 and user_name:
        rbac_note = (
            f"IMPORTANT: This user is an Employee named '{user_name}'. "
            f"Only return their own data (filter by AssignedTo or RecipientName LIKE '%{user_name.split()[0]}%')."
        )
    elif user_role == 2:
        rbac_note = 'This user is a Manager. They can see all team data.'
    else:
        rbac_note = 'This user is an Admin. They can see everything.'

    prompt = f"""
You are a SQL expert for the TaskTune AI work management system.
Today's date: {datetime.now().strftime('%Y-%m-%d')}
Current user role: {role_label}
{rbac_note}

{VIEWS_SCHEMA}

Rules:
1. ONLY use SELECT queries â€” never INSERT, UPDATE, DELETE, DROP.
2. ONLY query from: TaskFull, EvaluationView, NotificationsView.
3. For name searches, always use LIKE '%name%' (case-insensitive fuzzy match).
4. For 'overdue' tasks: WHERE DeadlineStatus = 'Overdue'.
5. For 'due soon': WHERE DeadlineStatus = 'DueSoon'.
6. Return ONLY the raw SQL query â€” no explanation, no markdown, no backticks.

Question: {question}
"""

    resp = client.chat.completions.create(
        model=GROQ_MODEL,
        messages=[{'role': 'user', 'content': prompt}],
        temperature=0,
    )
    sql = resp.choices[0].message.content.strip()
    sql = sql.replace('```sql', '').replace('```', '').strip()
    rows, err = run_sql(sql)
    return rows, sql, err


def format_sql_answer(question: str, rows: list, sql: str, user_role: int = 1) -> str:
    """Turn SQL rows into a natural-language answer (matches question language)."""
    if not rows:
        return 'No data found for your query.'
    client = _get_groq()
    prompt = f"""
You are TaskTune AI assistant.
Question: {question}
SQL results ({len(rows)} rows): {json.dumps(rows[:15], indent=2, default=str)}

Instructions:
- Answer in the EXACT same language as the question (Arabic/English/Mixed).
- Be concise and clear.
- Format dates as DD-MM-YYYY.
- Use bullet points for lists.
- NEVER invent data not in the results.
"""
    resp = client.chat.completions.create(
        model=GROQ_MODEL,
        messages=[{'role': 'user', 'content': prompt}],
    )
    return resp.choices[0].message.content


def evaluator_agent(
    stage: str,
    input_data,
    output_data,
    context: str = '',
    verbose: bool = False,
) -> dict:
    """
    QA evaluator after each pipeline stage.
    Returns dict with keys: score (0-10), verdict (pass|warn|fail), feedback.
    """
    client = _get_groq()
    prompt = f"""
You are a strict QA evaluator for a multi-agent AI system.

Stage: {stage}
Input: {str(input_data)[:400]}
Output: {str(output_data)[:600]}
Context: {str(context)[:200]}

Evaluate:
- Correctness: Is the output accurate for this stage?
- Completeness: Does it fully address the input?
- Format: Is it in the right format?

Respond ONLY in this JSON format:
{{"score": <0-10>, "verdict": "pass|warn|fail", "feedback": "<one sentence>"}}
"""
    resp = client.chat.completions.create(
        model=GROQ_MODEL,
        messages=[{'role': 'user', 'content': prompt}],
        temperature=0,
    )
    text = resp.choices[0].message.content.strip()
    try:
        result = json.loads(text[text.find('{') : text.rfind('}') + 1])
    except Exception:
        result = {'score': 5, 'verdict': 'warn', 'feedback': text[:100]}
    if verbose:
        icon = {'pass': 'âœ…', 'warn': 'âš ï¸', 'fail': 'âŒ'}.get(result.get('verdict', ''), 'â“')
        print(
            f"  {icon} Evaluator [{stage}] â€” {result.get('score', '?')}/10 | "
            f"{str(result.get('verdict', '')).upper()} | {result.get('feedback', '')}"
        )
    return result


def format_evaluations_summary(evals: list[dict]) -> str:
    """Compact one-line summary for chat UI (English labels)."""
    if not evals:
        return ''
    parts = []
    for ev in evals:
        v  = ev.get('verdict', '?')
        s  = ev.get('score', '?')
        fb = (ev.get('feedback') or '')[:120]
        parts.append(f'{v} {s}/10 â€” {fb}')
    return 'Evaluator | ' + ' Â· '.join(parts)


def ask_sql(
    question: str,
    user_role: int = 3,
    user_name: str | None = None,
    verbose: bool = False,
    use_evaluator: bool = True,
) -> tuple[str, str | None, str | None, list[dict]]:
    """
    Full SQL pipeline: NL â†’ SQL â†’ run â†’ (optional) evaluator stages â†’ natural-language answer.
    Returns (answer_text, sql_or_none, error_or_none, evaluations).
    """
    rows, sql, err = sql_agent(question, user_role, user_name)
    evaluations: list[dict] = []

    if err:
        if use_evaluator:
            evaluations.append(
                evaluator_agent('sql_generation', question, sql, err, verbose=verbose)
            )
        if verbose:
            print(f'SQL error: {err}\nSQL was: {sql}')
        return f'Sorry, I could not query the database: {err}', sql, err, evaluations

    if use_evaluator:
        evaluations.append(
            evaluator_agent(
                'sql_generation',
                question,
                {'sql': sql, 'rows': rows[:3]},
                '',
                verbose=verbose,
            )
        )

    answer = format_sql_answer(question, rows, sql or '', user_role)

    if use_evaluator:
        evaluations.append(
            evaluator_agent(
                'answer_quality',
                question,
                answer,
                json.dumps(rows[:3], default=str),
                verbose=verbose,
            )
        )

    if verbose:
        print(f'{len(rows)} rows | SQL: {sql}')

    return answer, sql, None, evaluations
````

## ðŸ“š Section 4 â€” RAG Layer

````python
# =============================================================================
# RAG layer
# =============================================================================

SUPPORTED_EXTENSIONS = {
    '.pdf', '.docx', '.doc', '.xlsx', '.xls', '.csv', '.txt',
    '.png', '.jpg', '.jpeg', '.webp',
}


def validate_file(path: str):
    if not os.path.exists(path):
        return False, f'File not found: {path}'
    if os.path.getsize(path) == 0:
        return False, f'File is empty: {path}'
    ext = os.path.splitext(path)[1].lower()
    if ext not in SUPPORTED_EXTENSIONS:
        return False, f"Unsupported type '{ext}'"
    return True, 'ok'


def validate_url(url: str):
    if not url.startswith(('http://', 'https://')):
        return False, f'Invalid URL: {url}'
    return True, 'ok'


def validate_knowledge_base(coll, emb: SentenceTransformer):
    errors = []
    if coll.count() == 0:
        errors.append('Knowledge base is empty')
        return False, errors
    try:
        e = emb.encode(['test']).tolist()
        results = coll.query(query_embeddings=e, n_results=1)
        if not results['documents'][0]:
            errors.append('Retrieval returned no results')
    except Exception as e:
        errors.append(f'Retrieval error: {e}')
    sample = coll.get(limit=5)['metadatas']
    for meta in sample:
        for key in ['source', 'page', 'type']:
            if key not in meta:
                errors.append(f"Missing metadata key '{key}'")
                break
    if errors:
        return False, errors
    all_m = coll.get()['metadatas']
    types   = Counter(m['type'] for m in all_m)
    sources = Counter(m['source'] for m in all_m)
    print(f'  Validation passed! {coll.count()} chunks | Types: {dict(types)}')
    for src, cnt in sources.most_common(5):
        print(f'    {src}: {cnt} chunks')
    return True, []


def extract_image_ocr(pil_img: Image.Image) -> str:
    """OCR â€” Arabic + English."""
    try:
        text = pytesseract.image_to_string(pil_img, lang='ara+eng')
        return text.strip()
    except Exception:
        return pytesseract.image_to_string(pil_img).strip()


def extract_pdf(path: str) -> list[dict]:
    """PDF: text, image OCR, tables per page."""
    chunks = []
    filename = os.path.basename(path)
    doc = fitz.open(path)
    for page_num, page in enumerate(doc, 1):
        text = page.get_text().strip()
        if text:
            chunks.append({'text': text, 'source': filename, 'page': page_num, 'type': 'text'})
        for img in page.get_images():
            try:
                img_bytes = doc.extract_image(img[0])['image']
                pil_img   = Image.open(io.BytesIO(img_bytes))
                ocr_text  = extract_image_ocr(pil_img)
                if ocr_text:
                    chunks.append({'text': f'[Image on page {page_num}]: {ocr_text}',
                                   'source': filename, 'page': page_num, 'type': 'image'})
            except Exception:
                pass
    with pdfplumber.open(path) as pdf:
        for page_num, page in enumerate(pdf.pages, 1):
            for table in page.extract_tables() or []:
                if not table or not table[0]:
                    continue
                header   = ' | '.join(str(c or '') for c in table[0])
                sep      = ' | '.join(['---'] * len(table[0]))
                rows_str = ['\n'.join([header, sep]) + '\n' + '\n'.join(
                    ' | '.join(str(c or '') for c in row) for row in table[1:])]
                chunks.append({'text': f'[Table on page {page_num}]:\n{rows_str[0]}',
                               'source': filename, 'page': page_num, 'type': 'table'})
    return chunks


def extract_docx(path: str) -> list[dict]:
    doc  = DocxDoc(path)
    text = '\n'.join(p.text for p in doc.paragraphs if p.text.strip())
    return [{'text': text, 'source': os.path.basename(path), 'page': 1, 'type': 'text'}]


def extract_excel(path: str) -> list[dict]:
    dfs   = pd.read_excel(path, sheet_name=None)
    parts = [f'Sheet: {s}\n{df.to_string(index=False)}' for s, df in dfs.items()]
    return [{'text': '\n\n'.join(parts), 'source': os.path.basename(path), 'page': 1, 'type': 'text'}]


def extract_csv(path: str) -> list[dict]:
    text = pd.read_csv(path).to_string(index=False)
    return [{'text': text, 'source': os.path.basename(path), 'page': 1, 'type': 'text'}]


def extract_txt(path: str) -> list[dict]:
    with open(path, encoding='utf-8', errors='ignore') as f:
        text = f.read()
    return [{'text': text, 'source': os.path.basename(path), 'page': 1, 'type': 'text'}]


def extract_image_file(path: str) -> list[dict]:
    ocr_text = extract_image_ocr(Image.open(path))
    return [{'text': ocr_text or f'[Image: {os.path.basename(path)}]',
             'source': os.path.basename(path), 'page': 1, 'type': 'image'}]


EXTRACTORS = {
    '.pdf':  extract_pdf,
    '.docx': extract_docx, '.doc': extract_docx,
    '.xlsx': extract_excel, '.xls': extract_excel,
    '.csv':  extract_csv,
    '.txt':  extract_txt,
    '.png':  extract_image_file, '.jpg': extract_image_file,
    '.jpeg': extract_image_file, '.webp': extract_image_file,
}


def extract_file(path: str) -> list[dict]:
    ok, msg = validate_file(path)
    if not ok:
        print(f'  Skipping: {msg}')
        return []
    return EXTRACTORS[os.path.splitext(path)[1].lower()](path)


def file_hash(path: str) -> str:
    with open(path, 'rb') as f:
        return hashlib.md5(f.read()).hexdigest()


def semantic_split(items: list[dict]) -> list[dict]:
    """Semantic chunking; keep tables/images small chunks unsplit."""
    result       = []
    lc_embeddings = HuggingFaceEmbeddings(model_name=EMBED_MODEL)
    chunker = SemanticChunker(
        lc_embeddings,
        breakpoint_threshold_type='percentile',
        breakpoint_threshold_amount=85,
    )
    for item in items:
        if item['type'] in ('table', 'image') or len(item['text'].split()) < 50:
            result.append(item)
            continue
        try:
            docs = chunker.create_documents([item['text']])
            for doc in docs:
                result.append({'text': doc.page_content, 'source': item['source'],
                                'page': item['page'], 'type': item['type']})
        except Exception:
            result.append(item)
    return result


async def scrape(url: str, label: str, wait: int = 2000) -> str:
    """Playwright scrape (JS-friendly)."""
    ok, msg = validate_url(url)
    if not ok:
        print(f'  {msg}')
        return ''
    try:
        async with async_playwright() as p:
            browser = await p.chromium.launch()
            page    = await browser.new_page()
            await page.goto(url, timeout=30000, wait_until='domcontentloaded')
            await page.wait_for_timeout(wait)
            text = await page.inner_text('body')
            await browser.close()
            lines = [ln for ln in text.splitlines() if len(ln.strip()) > 30]
            return '\n'.join(lines[:400])
    except Exception as e:
        print(f'  {label}: {e}')
        return ''


def scrape_sync(url: str, label: str) -> str:
    """Sync wrapper â€” isolated event loop for Gradio compatibility."""
    loop = asyncio.new_event_loop()
    asyncio.set_event_loop(loop)
    try:
        return loop.run_until_complete(scrape(url, label))
    finally:
        loop.close()


def build_knowledge_base(extra_text: str | None = None) -> int:
    """Download PDFs, extract files, optional web scrape, chunk, embed, store in Chroma."""
    global embedder, collection, WEBSITES

    all_items: list[dict] = []
    processed_hashes: set[str] = set()

    for filename, url in PDF_URLS.items():
        if not os.path.exists(filename):
            ok, msg = validate_url(url)
            if not ok:
                print(f'  Skipping: {msg}')
                continue
            print(f'  Downloading {filename}...')
            r = requests.get(url, timeout=60, headers={'User-Agent': 'Mozilla/5.0'})
            if r.status_code == 200:
                open(filename, 'wb').write(r.content)
                print(f'  Saved {filename}')
            else:
                print(f'  HTTP {r.status_code} for {filename}')

    patterns = ['*.pdf','*.docx','*.doc','*.xlsx','*.xls','*.csv','*.txt',
                '*.png','*.jpg','*.jpeg','*.webp']
    all_files = [f for p in patterns for f in glob.glob(p)]
    print(f'Found {len(all_files)} file(s)')
    for f in all_files:
        fh = file_hash(f)
        if fh in processed_hashes:
            print(f'  Duplicate content skipped: {os.path.basename(f)}')
            continue
        try:
            items = extract_file(f)
            all_items.extend(items)
            processed_hashes.add(fh)
        except Exception as e:
            print(f'  {f}: {e}')

    if WEBSITES:
        print(f'Scraping {len(WEBSITES)} site(s)...')
        for label, url in WEBSITES.items():
            text = scrape_sync(url, label)
            if len(text) > 200:
                all_items.append({'text': text, 'source': label, 'page': 1, 'type': 'web'})

    if extra_text and extra_text.strip():
        all_items.append({'text': extra_text, 'source': 'Manual input', 'page': 1, 'type': 'text'})

    if not all_items:
        print('No content to index.')
        return 0

    print(f'Semantic chunking {len(all_items)} items...')
    chunks = semantic_split(all_items)
    print(f'  -> {len(chunks)} chunks')

    print('Embedding...')
    embedder = SentenceTransformer(EMBED_MODEL)
    chroma   = chromadb.Client()
    try:
        chroma.delete_collection('rag_db')
    except Exception:
        pass
    collection = chroma.create_collection('rag_db')
    texts = [c['text'] for c in chunks]
    metas = [{'source': c['source'], 'page': str(c['page']), 'type': c['type']} for c in chunks]
    embs  = embedder.encode(texts, show_progress_bar=True).tolist()
    collection.add(
        ids=[str(i) for i in range(len(texts))],
        embeddings=embs,
        documents=texts,
        metadatas=metas,
    )
    ok, _ = validate_knowledge_base(collection, embedder)
    return collection.count() if ok else 0


def search(question: str, n: int = 10):
    """Vector search in Chroma."""
    if collection is None or embedder is None:
        return [], []
    emb = embedder.encode([question]).tolist()
    res = collection.query(query_embeddings=emb, n_results=n)
    return res['documents'][0], res['metadatas'][0]


def retrieve(question: str):
    """Bilingual retrieval: query + English paraphrase."""
    client = _get_groq()
    chunks_a, metas_a = search(question, n=10)
    t = client.chat.completions.create(
        model=GROQ_MODEL,
        messages=[{'role': 'user', 'content': f'Translate to English, one line only: {question}'}],
    )
    en_q = t.choices[0].message.content.strip()
    chunks_b, metas_b = search(en_q, n=10)
    seen, final_chunks, final_metas = set(), [], []
    for chunk, meta in zip(chunks_a + chunks_b, metas_a + metas_b):
        if chunk not in seen:
            seen.add(chunk)
            final_chunks.append(chunk)
            final_metas.append(meta)
    return final_chunks[:11], final_metas[:11]


def evaluate_rag(question: str, answer: str, context: str) -> str:
    """Self-RAG: one-word quality check."""
    client = _get_groq()
    prompt = (
        f'Question: {question}\n'
        f'Context: {context[:1000]}\n'
        f'Answer: {answer}\n\n'
        'Is this answer supported by the context? Reply with one word: good / partial / bad'
    )
    res = client.chat.completions.create(
        model=GROQ_MODEL, messages=[{'role': 'user', 'content': prompt}]
    )
    verdict = res.choices[0].message.content.strip().lower()
    if 'good'    in verdict: return 'good'
    if 'partial' in verdict: return 'partial'
    return 'bad'


def format_sources(metas: list) -> str:
    """Citation line: file + page + type icon."""
    seen, parts = set(), []
    for meta in metas:
        key = f"{meta['source']}|{meta['page']}"
        if key not in seen:
            seen.add(key)
            icon = {'text': 'ðŸ“„', 'image': 'ðŸ–¼ï¸', 'table': 'ðŸ“Š', 'web': 'ðŸŒ'}.get(
                meta.get('type', ''), 'ðŸ“„'
            )
            parts.append(f"{icon} {meta['source']} p.{meta['page']}")
    return ' | '.join(parts)


def ask(
    question: str, max_retries: int = 2
) -> tuple[str, str, str, str]:
    """
    Self-RAG pipeline. Returns (answer, sources_line, verdict, partial_note).
    """
    global memory
    if not question.strip():
        return '', '', 'bad', ''
    if collection is None or embedder is None:
        return (
            'Knowledge base not built yet. Use **Build Knowledge Base** first.',
            '', 'bad', '',
        )

    client = _get_groq()
    answer = ''
    for attempt in range(max_retries):
        chunks, metas = retrieve(question)
        context = '\n\n'.join(chunks)
        messages = [
            {
                'role': 'system',
                'content': (
                    'You are a helpful assistant. '
                    'Answer ONLY from the provided context â€” do not use outside knowledge. '
                    'Respond in the EXACT same language as the question. '
                    'Arabic question â†’ Arabic ONLY, no other language. '
                    'English question â†’ English ONLY. '
                    'Mixed â†’ mirror the exact same mix. '
                    'NEVER use Vietnamese, Thai, Russian, or any other language. '
                    'Write clean flowing text â€” no \\n escape characters, use natural paragraph breaks only. '
                    'Skip garbled or unreadable text silently. '
                    'If the answer is not in the context, say so in one sentence.'
                ),
            }
        ]
        for h in memory[-4:]:
            messages.append({'role': 'user',      'content': h['q']})
            messages.append({'role': 'assistant', 'content': h['a']})
        messages.append({'role': 'user', 'content': f'Context:\n{context}\n\nQuestion: {question}'})

        resp   = client.chat.completions.create(model=GROQ_MODEL, messages=messages)
        answer = resp.choices[0].message.content
        verdict = evaluate_rag(question, answer, context)

        if verdict in ('good', 'partial') or attempt == max_retries - 1:
            sources = format_sources(metas)
            note    = ' *(partial â€” some details may be missing)*' if verdict == 'partial' else ''
            memory.append({'q': question, 'a': answer})
            if len(memory) > 10:
                memory = memory[-10:]
            return answer, sources, verdict, note

    return answer, '', 'bad', ''


def clear_memory():
    global memory
    memory = []
    print('Memory cleared')
````

## ðŸ”€ Section 6 â€” Smart Router + Memory + LangGraph

**New Flow:**
- **Understand Layer** â†’ intent (`chitchat` / `vague` / `real_question`) + complexity + freshness
- **Decision Gate** â†’ chitchat/vague go to General LLM immediately
- **Planner** â†’ decides which sources (sql / rag / web) and how many
- **Retrieval** â†’ sequential (simple/medium) or **parallel** (complex)
- **Verification** â†’ checks result quality
- **Smart Retry** â†’ tries untried sources before giving up (max 2 retries)
- **Smart Merge** â†’ single source returns directly; multi-source uses LLM synthesis with priority SQL > RAG > Web
- **General LLM Fallback** â†’ last resort when no data found


````python
from __future__ import annotations

import asyncio
import json as _json
import re
from typing import Annotated, Literal, Sequence, TypedDict

from langchain_core.messages import AIMessage, BaseMessage, HumanMessage
from langgraph.graph import END, START, StateGraph
from langgraph.graph.message import add_messages


# =============================================================================
# Memory â€” last N messages + rolling summary
# =============================================================================

MEMORY_K = 6
_conv_memory: list[dict] = []
_conv_summary: str = ""


def _update_memory(q: str, a: str, route: str) -> None:
    """Add a new turn; compress oldest into summary when exceeding MEMORY_K."""
    global _conv_memory, _conv_summary
    _conv_memory.append({"q": q, "a": a, "route": route})
    if len(_conv_memory) > MEMORY_K:
        to_compress = _conv_memory[:-MEMORY_K]
        _conv_memory = _conv_memory[-MEMORY_K:]
        client = _get_groq()
        old_block = "\n".join(
            f"User: {m['q']}\nAssistant: {m['a']}" for m in to_compress
        )
        prev = f"Previous summary:\n{_conv_summary}\n\n" if _conv_summary else ""
        r = client.chat.completions.create(
            model=GROQ_MODEL,
            messages=[{
                "role": "user",
                "content": (
                    f"{prev}Summarize the following conversation turns in 2-3 sentences, "
                    "keeping key facts and decisions. Same language as conversation.\n\n"
                    'Mixed â†’ mirror the exact same mix. '
                    'NEVER use Vietnamese, Thai, Russian, or any other language. '
                    'Write clean flowing text â€” no \\n escape characters, use natural paragraph breaks only. '
                    f"{old_block}"
                ),
            }],
            temperature=0,
            max_tokens=200,
        )
        _conv_summary = r.choices[0].message.content.strip()


def get_memory_context() -> str:
    """Build context string from rolling summary + last K messages."""
    parts = []
    if _conv_summary:
        parts.append(f"[Conversation summary so far]:\n{_conv_summary}")
    for m in _conv_memory:
        parts.append(f"User: {m['q']}\nAssistant: {m['a']}")
    return "\n\n".join(parts)


def clear_memory() -> None:
    global _conv_memory, _conv_summary
    _conv_memory = []
    _conv_summary = ""


# =============================================================================
# RBAC context (set by Gradio before each invoke)
# =============================================================================

_runtime_user_role: int = 3
_runtime_user_name: str | None = None


def set_rbac_context(role: int, name: str | None) -> None:
    global _runtime_user_role, _runtime_user_name
    _runtime_user_role = int(role)
    _runtime_user_name = name.strip() if name else None


# =============================================================================
# Graph State
# =============================================================================

class GraphState(TypedDict):
    messages:     Annotated[Sequence[BaseMessage], add_messages]

    # Understand Layer output
    intent:       str    # "chitchat" | "vague" | "real_question"
    complexity:   str    # "simple" | "medium" | "complex"
    freshness:    bool   # True â†’ needs real-time / web data

    # Planner output
    sources:      list   # subset of ["sql", "rag", "web"]

    # Retrieval results
    rag_result:   str
    sql_result:   str
    web_result:   str
    retry_count:  int    # guard against infinite retry loops

    # Final
    final:        str
    meta:         str


def _last_human(state: GraphState) -> str:
    for m in reversed(state["messages"]):
        if isinstance(m, HumanMessage):
            return m.content if isinstance(m.content, str) else str(m.content)
    return ""


# =============================================================================
# Node 1 â€” Understand Layer
# Extracts intent + complexity + freshness in ONE LLM call
# =============================================================================

_UNDERSTAND_PROMPT = """You are a query analyst for TaskTune AI (work management + document Q&A).

Analyze the user message and return a JSON object with EXACTLY these three fields:

{
  "intent":     "chitchat" | "vague" | "real_question",
  "complexity": "simple"   | "medium" | "complex",
  "freshness":  true | false
}

Definitions:
- intent:
    "chitchat"      â†’ greetings, thanks, small-talk, compliments, empty noise
    "vague"         â†’ under-specified, missing key context (e.g. "tell me something")
    "real_question" â†’ anything with a clear, answerable intent

- complexity:
    "simple"   â†’ one clear fact, one source is enough
    "medium"   â†’ needs two sources or moderate reasoning
    "complex"  â†’ multi-part, cross-source, or requires synthesis of 3+ sources

- freshness:
    true  â†’ involves current events, today's prices, latest versions, real-time data, news
    false â†’ otherwise

Return ONLY the JSON object. No markdown, no explanation.
Handle English and Arabic messages correctly.
'Respond in the EXACT same language as the question. '
'Arabic question â†’ Arabic ONLY, no other language. '
'English question â†’ English ONLY. '
'Mixed â†’ mirror the exact same mix. '
'NEVER use Vietnamese, Thai, Russian, or any other language. '
'Write clean flowing text â€” no \\n escape characters, use natural paragraph breaks only. '
                    """


def node_understand(state: GraphState) -> dict:
    text   = _last_human(state)
    client = _get_groq()
    try:
        r = client.chat.completions.create(
            model=GROQ_MODEL,
            messages=[
                {"role": "system", "content": _UNDERSTAND_PROMPT},
                {"role": "user",   "content": text.strip()},
            ],
            temperature=0,
            max_tokens=80,
        )
        raw = re.sub(r"```[a-z]*", "", r.choices[0].message.content or "").strip().strip("`").strip()
        parsed     = _json.loads(raw)
        intent     = parsed.get("intent",     "real_question")
        complexity = parsed.get("complexity", "simple")
        freshness  = bool(parsed.get("freshness", False))
    except Exception:
        intent, complexity, freshness = "real_question", "simple", False

    # keyword safety-nets
    t = text.lower()
    if re.search(r"\b(hi|hello|hey|thanks|Ù…Ø±Ø­Ø¨Ø§|Ø´ÙƒØ±Ø§Ù‹|Ø´ÙƒØ±|Ø£Ù‡Ù„Ø§|Ù‡Ø§ÙŠ)\b", t):
        intent = "chitchat"
    if re.search(r"\b(today|now|latest|current|recent|news|price|Ø³Ø¹Ø±|Ø§Ù„ÙŠÙˆÙ…|Ø§Ù„Ø¢Ù†|Ø£Ø­Ø¯Ø«|Ø£Ø®Ø¨Ø§Ø±|Ø¯Ù„ÙˆÙ‚ØªÙŠ|Ø§Ù„Ù†Ù‡Ø§Ø±Ø¯Ù‡)\b", t):
        freshness = True

    return {
        "intent":     intent,
        "complexity": complexity,
        "freshness":  freshness,
        "meta":       f"intent={intent}|complexity={complexity}|freshness={freshness}",
    }


def edge_after_understand(state: GraphState) -> str:
    intent = state.get("intent", "real_question")
    if intent == "chitchat":    return "general_chitchat"
    if intent == "vague":       return "general_clarify"
    return "planner"


# =============================================================================
# Nodes 2a/2b/2c â€” General LLM (three distinct roles)
# =============================================================================

_BASE_LLM_SYSTEM = (
    "Respond in the EXACT same language as the question. "
    "Arabic question â†’ Arabic ONLY. English â†’ English ONLY. "
    "Mixed â†’ mirror the exact mix. "
    'Respond in the EXACT same language as the question. '
    'NEVER use Vietnamese, Thai, Russian, or any other language. '
    "Write clean flowing text â€” no \\n escape characters. "
    "Skip garbled or unreadable text silently."
)


def _general_llm_call(system_extra: str, text: str, temperature: float = 0.4) -> str:
    client = _get_groq()
    ctx    = get_memory_context()
    system = _BASE_LLM_SYSTEM + "\n\n" + system_extra
    if ctx:
        system += f"\n\nConversation context:\n{ctx}"
    r = client.chat.completions.create(
        model=GROQ_MODEL,
        messages=[
            {"role": "system", "content": system},
            {"role": "user",   "content": text},
        ],
        temperature=temperature,
    )
    return (r.choices[0].message.content or "").strip()


def node_general_chitchat(state: GraphState) -> dict:
    """ðŸ’¬ Friendly chit-chat reply."""
    text = _last_human(state)
    ans  = _general_llm_call(
        "You are TaskTune AI. Reply warmly and briefly. No database or document lookup.",
        text,
    )
    _update_memory(text, ans, "general_chitchat")
    return {"messages": [AIMessage(content=ans)], "final": ans, "meta": "route=general_chitchat"}


def node_general_clarify(state: GraphState) -> dict:
    """â“ Ask ONE clarifying question for vague inputs."""
    text = _last_human(state)
    ans  = _general_llm_call(
        ("You are TaskTune AI. The user's question is under-specified. "
         "Ask ONE concise clarifying question to understand what they need. "
         "Do not attempt to answer yet."),
        text, temperature=0.3,
    )
    _update_memory(text, ans, "general_clarify")
    return {"messages": [AIMessage(content=ans)], "final": ans, "meta": "route=general_clarify"}


def node_general_fallback(state: GraphState) -> dict:
    """ðŸ›Ÿ No data found â€” answer from LLM knowledge as last resort."""
    text = _last_human(state)
    ans  = _general_llm_call(
        ( "You are TaskTune AI. No database or document results were found. "
          "Answer from your general knowledge as best you can. "
          'Respond in the EXACT same language as the question. '
          'Arabic question â†’ Arabic ONLY, no other language. '
          'English question â†’ English ONLY. '
          'Mixed â†’ mirror the exact same mix. '
          'NEVER use Vietnamese, Thai, Russian, or any other language. '
          'Write clean flowing text â€” no \\n escape characters, use natural paragraph breaks only. '
         "If you are not confident, say so briefly."),
        text, temperature=0.4,
    )
    _update_memory(text, ans, "general_fallback")
    badge   = "ðŸ§  llm-fallback"
    content = f"{ans}\n\n{badge}"
    return {"messages": [AIMessage(content=content)], "final": content, "meta": "route=general_fallback"}


# =============================================================================
# Node 3 â€” Planner
# Decides WHICH sources to use and HOW MANY
# =============================================================================

_PLANNER_PROMPT = """You are a retrieval planner for TaskTune AI.

Given a user question, decide which data sources are needed.
Available sources: sql, rag, web

- sql  â†’ counts, lists, tasks, deadlines, evaluations, notifications â€” structured DB data
- rag  â†’ policies, documents, explanations, unstructured knowledge
- web  â†’ real-time, current events, today's prices, latest news

Return a JSON array with the sources to use, ordered by priority.
Examples:
  "how many tasks are overdue?"                 â†’ ["sql"]
  "what does the leave policy say?"             â†’ ["rag"]
  "latest AI news"                              â†’ ["web"]
  "compare our policy vs industry standards"   â†’ ["rag", "web"]
  "summarize evaluations and explain the KPIs" â†’ ["sql", "rag"]
  "complex cross-source question"              â†’ ["sql", "rag", "web"]

Return ONLY the JSON array. No markdown, no explanation."""


def node_planner(state: GraphState) -> dict:
    text       = _last_human(state)
    complexity = state.get("complexity", "simple")
    freshness  = state.get("freshness",  False)
    client     = _get_groq()

    try:
        r = client.chat.completions.create(
            model=GROQ_MODEL,
            messages=[
                {"role": "system", "content": _PLANNER_PROMPT},
                {"role": "user",   "content": text.strip()},
            ],
            temperature=0, max_tokens=40,
        )
        raw     = re.sub(r"```[a-z]*", "", r.choices[0].message.content or "").strip()
        sources = _json.loads(raw)
        sources = [s for s in sources if s in ("sql", "rag", "web")]
    except Exception:
        sources = []

    # keyword safety-net
    if not sources:
        t = text.lower()
        if re.search(r"\b(count|how many|overdue|task|deadline|evaluation|notification)\b", t):
            sources = ["sql"]
        elif freshness:
            sources = ["web"]
        else:
            sources = ["rag"]

    # force web when freshness=True and complex
    if freshness and "web" not in sources and complexity == "complex":
        sources.append("web")

    # ensure 2+ sources for complex
    if complexity == "complex" and len(sources) < 2:
        sources = (["sql"] + sources) if "sql" not in sources else (sources + ["rag"])

    return {
        "sources":     sources,
        "rag_result":  "",
        "sql_result":  "",
        "web_result":  "",
        "retry_count": 0,
        "meta":        f"sources={sources}",
    }


# =============================================================================
# Node 4 â€” Retrieval Layer
# Runs planned sources (parallel for complex, sequential otherwise)
# =============================================================================

def _run_rag(text: str) -> str:
    try:
        ans, sources, verdict, note = ask(text)
        if verdict == "bad" or not ans:
            return ""
        bad = ("not in the context", "Ù„Ù… Ø£Ø¬Ø¯", "Ù„Ø§ ØªÙˆØ¬Ø¯", "no information")
        if any(p in ans.lower() for p in bad):
            return ""
        return ans.strip()
    except Exception:
        return ""


def _run_sql(text: str) -> str:
    try:
        ans, _sql, err, evals = ask_sql(
            text,
            user_role=_runtime_user_role,
            user_name=_runtime_user_name,
            verbose=False,
            use_evaluator=True,
        )
        if err:
            return ""
        bad = ("no data found", "could not query", "no results")
        if any(p in (ans or "").lower() for p in bad):
            return ""
        footer = format_evaluations_summary(evals)
        return (ans + (f"\n\n_{footer}_" if footer else "")).strip()
    except Exception:
        return ""


def _run_web(text: str) -> str:
    try:
        return (ask_web(text) or "").strip()
    except Exception:
        return ""


def node_retrieval(state: GraphState) -> dict:
    text       = _last_human(state)
    sources    = state.get("sources", ["rag"])
    complexity = state.get("complexity", "simple")

    rag_result = state.get("rag_result", "")
    sql_result = state.get("sql_result", "")
    web_result = state.get("web_result", "")

    if complexity == "complex" and len(sources) > 1:
        # Parallel execution
        import concurrent.futures
        with concurrent.futures.ThreadPoolExecutor() as pool:
            futures = {}
            if "rag" in sources: futures["rag"] = pool.submit(_run_rag, text)
            if "sql" in sources: futures["sql"] = pool.submit(_run_sql, text)
            if "web" in sources: futures["web"] = pool.submit(_run_web, text)
            results = {k: f.result() for k, f in futures.items()}
        rag_result = results.get("rag", "")
        sql_result = results.get("sql", "")
        web_result = results.get("web", "")
    else:
        # Sequential
        for src in sources:
            if src == "rag" and not rag_result: rag_result = _run_rag(text)
            elif src == "sql" and not sql_result: sql_result = _run_sql(text)
            elif src == "web" and not web_result: web_result = _run_web(text)

    return {"rag_result": rag_result, "sql_result": sql_result, "web_result": web_result}


# =============================================================================
# Node 5 â€” Verification Layer
# =============================================================================

MAX_RETRIES = 2


def node_verify(state: GraphState) -> dict:
    return {}  # routing logic lives in the edge


def edge_after_verify(state: GraphState) -> str:
    has_data = bool(
        (state.get("rag_result") or "").strip() or
        (state.get("sql_result") or "").strip() or
        (state.get("web_result") or "").strip()
    )
    if has_data:
        return "merge_and_answer"
    if state.get("retry_count", 0) < MAX_RETRIES:
        return "smart_retry"
    return "general_fallback"


# =============================================================================
# Node 6 â€” Smart Retry
# Adds one untried source and re-runs retrieval
# =============================================================================

def node_smart_retry(state: GraphState) -> dict:
    tried   = state.get("sources", [])
    untried = [s for s in ["sql", "rag", "web"] if s not in tried]
    new_src = tried + (untried[:1] if untried else ["web"])
    return {
        "sources":     new_src,
        "rag_result":  "",
        "sql_result":  "",
        "web_result":  "",
        "retry_count": state.get("retry_count", 0) + 1,
        "meta":        f"retry#{state.get('retry_count',0)+1}|sources={new_src}",
    }


# =============================================================================
# Node 7 â€” Smart Merge + Answer Generator
# Priority: SQL > RAG > Web
# =============================================================================

def node_merge_and_answer(state: GraphState) -> dict:
    text = _last_human(state)
    sql  = (state.get("sql_result") or "").strip()
    rag  = (state.get("rag_result") or "").strip()
    web  = (state.get("web_result") or "").strip()

    filled = [(lbl, cnt) for lbl, cnt in
              [("Database", sql), ("Documents", rag), ("Web", web)] if cnt]

    if not filled:
        return {"final": "", "meta": "merge=empty"}

    # Single source â€” return directly with badge
    if len(filled) == 1:
        label, content = filled[0]
        badge = {"Database": "ðŸ—„ï¸ sql", "Documents": "ðŸ“š rag", "Web": "ðŸŒ web"}.get(label, "")
        final = f"{content}\n\n{badge}"
        _update_memory(text, content, label.lower())
        return {"messages": [AIMessage(content=final)], "final": final,
                "meta": f"merge=single|source={label}"}

    # Multiple sources â€” LLM synthesis
    client = _get_groq()
    ctx    = get_memory_context()
    sources_block = "\n\n".join(f"--- {lbl} ---\n{cnt}" for lbl, cnt in filled)

    r = client.chat.completions.create(
        model=GROQ_MODEL,
        messages=[
            {
                "role": "system",
                "content": (
                    "You are TaskTune AI. Merge the provided source results into ONE "
                    "coherent answer in the user's language. "
                    'Respond in the EXACT same language as the question. '
                    'Arabic question â†’ Arabic ONLY, no other language. '
                    'English question â†’ English ONLY. '
                    'Mixed â†’ mirror the exact same mix. '
                    'NEVER use Vietnamese, Thai, Russian, or any other language. '
                    'Write clean flowing text â€” no \\n escape characters, use natural paragraph breaks only. '
                    "Priority: Database > Documents > Web. "
                    "Use clear sections if multiple sources contributed meaningfully. "
                    "If a source is partial, integrate what's useful and skip the rest. "
                    + _BASE_LLM_SYSTEM
                    + (f"\n\nConversation context:\n{ctx}" if ctx else "")
                ),
            },
            {"role": "user", "content": f"Question:\n{text}\n\n{sources_block}"},
        ],
        temperature=0.2,
    )
    merged = (r.choices[0].message.content or "").strip()
    badge  = " ".join(
        e for e in [
            "ðŸ—„ï¸" if any(l == "Database"  for l, _ in filled) else "",
            "ðŸ“š" if any(l == "Documents" for l, _ in filled) else "",
            "ðŸŒ" if any(l == "Web"       for l, _ in filled) else "",
        ] if e
    )
    labels = "+".join(l.lower()[:3] for l, _ in filled)
    final  = f"{merged}\n\n{badge}"
    _update_memory(text, merged, f"merge:{labels}")
    return {"messages": [AIMessage(content=final)], "final": final,
            "meta": f"merge=multi|sources={labels}"}


# =============================================================================
# Build LangGraph
# =============================================================================

def build_router_graph():
    g = StateGraph(GraphState)

    g.add_node("understand",         node_understand)
    g.add_node("general_chitchat",   node_general_chitchat)
    g.add_node("general_clarify",    node_general_clarify)
    g.add_node("general_fallback",   node_general_fallback)
    g.add_node("planner",            node_planner)
    g.add_node("retrieval",          node_retrieval)
    g.add_node("verify",             node_verify)
    g.add_node("smart_retry",        node_smart_retry)
    g.add_node("merge_and_answer",   node_merge_and_answer)

    g.add_edge(START, "understand")

    g.add_conditional_edges(
        "understand",
        edge_after_understand,
        {
            "general_chitchat": "general_chitchat",
            "general_clarify":  "general_clarify",
            "planner":          "planner",
        },
    )

    g.add_edge("general_chitchat", END)
    g.add_edge("general_clarify",  END)
    g.add_edge("planner",          "retrieval")
    g.add_edge("retrieval",        "verify")

    g.add_conditional_edges(
        "verify",
        edge_after_verify,
        {
            "merge_and_answer": "merge_and_answer",
            "smart_retry":      "smart_retry",
            "general_fallback": "general_fallback",
        },
    )

    g.add_edge("smart_retry",      "retrieval")
    g.add_edge("merge_and_answer", END)
    g.add_edge("general_fallback", END)

    return g.compile()


ROUTER_APP = build_router_graph()


# =============================================================================
# Public entry point â€” drop-in replacement for old chat_router
# =============================================================================

def chat_router(message: str) -> tuple[str, str]:
    """Returns (assistant_reply, meta_line)."""
    out = ROUTER_APP.invoke(
        {
            "messages":    [HumanMessage(content=message)],
            "intent":      "",
            "complexity":  "",
            "freshness":   False,
            "sources":     [],
            "rag_result":  "",
            "sql_result":  "",
            "web_result":  "",
            "retry_count": 0,
            "final":       "",
            "meta":        "",
        }
    )
    return out.get("final", ""), out.get("meta", "")
````

````python
# =============================================================================
# Gradio â€” combined UI with persistent chat history
# =============================================================================

CSS = """
body, .gradio-container { background: #0d0d1a !important; }

/* primary buttons â€” same gradient as header */
.gr-button-primary, button.primary {
    background: linear-gradient(135deg, #0f0c29, #302b63, #24243e) !important;
    border: 1px solid #4f46e5 !important;
    color: #c4b5fd !important;
    font-weight: 600 !important;
    box-shadow: 0 0 20px rgba(79,70,229,0.3) !important;
    transition: all 0.2s ease !important;
}
.gr-button-primary:hover, button.primary:hover {
    box-shadow: 0 0 35px rgba(79,70,229,0.6) !important;
    border-color: #818cf8 !important;
    color: white !important;
}

/* secondary buttons */
.gr-button-secondary, button.secondary {
    background: #1a1a2e !important;
    border: 1px solid #3d3d6e !important;
    color: #a78bfa !important;
    transition: all 0.2s ease !important;
}
.gr-button-secondary:hover, button.secondary:hover {
    border-color: #7c3aed !important;
    box-shadow: 0 0 10px rgba(124,58,237,0.25) !important;
}

/* text inputs */
.gr-textbox textarea, .gr-textbox input {
    background: #12121f !important;
    border: 1px solid #2d2d4e !important;
    color: #e2e8f0 !important;
    border-radius: 10px !important;
}
.gr-textbox textarea:focus, .gr-textbox input:focus {
    border-color: #7c3aed !important;
    box-shadow: 0 0 12px rgba(124,58,237,0.35) !important;
}

/* panels */
.gr-box, .gr-panel {
    background: #12121f !important;
    border: 1px solid #2d2d4e !important;
    border-radius: 16px !important;
}

/* labels */
label { color: #a78bfa !important; font-weight: 500 !important; }

/* section headings */
.gr-markdown h3 {
    color: #c4b5fd !important;
    border-bottom: 1px solid #2d2d4e !important;
    padding-bottom: 8px !important;
}

/* examples box container */
.examples-box {
    background: linear-gradient(135deg, #0f0c29, #1a1640) !important;
    border: 1px solid #4f46e5 !important;
    border-radius: 14px !important;
    padding: 14px !important;
    box-shadow: 0 0 20px rgba(79,70,229,0.15) !important;
    margin-bottom: 10px !important;
}

/* history sidebar */
.history-item {
    background: #1a1a2e;
    border: 1px solid #2d2d4e;
    border-radius: 10px;
    padding: 10px;
    margin-bottom: 8px;
    cursor: pointer;
    color: #c4b5fd;
    font-size: 13px;
}
.history-item:hover {
    border-color: #7c3aed;
    background: #20204a;
}
"""

# â”€â”€ in-memory chat history store â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# Structure: { conv_id: {"title": str, "messages": [...], "summary": str} }
_all_conversations: dict = {}
_active_conv_id: str = ""


def _new_conv_id() -> str:
    import time
    return f"conv_{int(time.time()*1000)}"


def _save_active(history: list, summary: str) -> None:
    """Persist current chat to the store."""
    global _active_conv_id
    if not _active_conv_id:
        return
    if _active_conv_id not in _all_conversations:
        _all_conversations[_active_conv_id] = {"title": "", "messages": [], "summary": ""}
    _all_conversations[_active_conv_id]["messages"] = history
    _all_conversations[_active_conv_id]["summary"] = summary
    # auto-title from first user message
    if not _all_conversations[_active_conv_id]["title"] and history:
        first_user = next((m["content"] for m in history if m["role"] == "user"), "")
        _all_conversations[_active_conv_id]["title"] = first_user[:40] or "New chat"


def _history_html() -> str:
    """Render sidebar history list."""
    if not _all_conversations:
        return "<p style=\'color:#6b7280;font-size:13px;\'>No history yet.</p>"
    items = []
    for cid, cv in reversed(list(_all_conversations.items())):
        title   = cv.get("title", "New chat")[:38]
        summary = cv.get("summary", "")[:80]
        active  = "border-color:#7c3aed !important;" if cid == _active_conv_id else ""
        items.append(
            f"<div class=\'history-item\' style=\'{active}\' "
            f"onclick=\'document.querySelector("#load_conv_id textarea").value="{cid}"; "
            f"document.querySelector("#load_conv_btn").click();\' >"
            f"<b>{title}</b>"
            + (f"<br><span style=\'color:#6b7280;font-size:12px;\'>{summary}</span>" if summary else "")
            + "</div>"
        )
    return "".join(items)


def launch_gradio(share: bool = True):
    import gradio as gr

    init_db()

    # â”€â”€ Knowledge base builder â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
    def process_and_build(files, urls_text, manual_text):
        global WEBSITES
        if files:
            for f in files:
                shutil.copy(f, os.path.basename(f))
        if urls_text and urls_text.strip():
            lines = [ln.strip() for ln in urls_text.strip().splitlines() if ln.strip()]
            WEBSITES = {
                f"Page {i+1} ({url.split('/')[2]})": url
                for i, url in enumerate(lines)
            }
        else:
            WEBSITES = {}
        count = build_knowledge_base(extra_text=manual_text)
        if count == 0:
            return "Build finished with 0 chunks â€” add files or text."
        return f"Indexed {count} chunks. Ready to chat."

    # â”€â”€ Send message â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
    def respond(message, history, role, emp_name):
        global _active_conv_id
        if not message or not str(message).strip():
            return "", history, _history_html()
        if not _active_conv_id:
            _active_conv_id = _new_conv_id()
        role_map = {"Admin": 3, "Manager": 2, "Employee": 1}
        set_rbac_context(role_map.get(role, 3), emp_name or None)
        try:
            reply, meta = chat_router(message)
            # badge already embedded by nodes; strip if empty
            if not reply.strip():
                reply = ("Ù„Ù… Ø£Ø¬Ø¯ Ø¥Ø¬Ø§Ø¨Ø© ÙƒØ§ÙÙŠØ©." if re.search(r"[\u0600-\u06FF]", message)
                         else "I couldn\'t find a sufficient answer.")
            history = history + [
                {"role": "user",      "content": message},
                {"role": "assistant", "content": reply},
            ]
        except Exception as e:
            history = history + [
                {"role": "user",      "content": message},
                {"role": "assistant", "content": f"Error: {e}"},
            ]
        _save_active(history, _conv_summary)
        return "", history, _history_html()

    # â”€â”€ New conversation â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
    def new_conv():
        global _active_conv_id
        _active_conv_id = _new_conv_id()
        clear_memory()
        return [], _history_html()

    # â”€â”€ Load conversation â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
    def load_conv(conv_id):
        global _active_conv_id
        cv = _all_conversations.get(conv_id.strip())
        if not cv:
            return gr.update(), _history_html()
        _active_conv_id = conv_id.strip()
        clear_memory()
        # rebuild memory from saved messages
        msgs = cv.get("messages", [])
        for i in range(0, len(msgs) - 1, 2):
            if msgs[i]["role"] == "user" and i+1 < len(msgs):
                _update_memory(msgs[i]["content"], msgs[i+1]["content"], "loaded")
        return cv.get("messages", []), _history_html()

    # â”€â”€ Edit conversation title â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
    def rename_conv(new_title):
        if _active_conv_id and _active_conv_id in _all_conversations:
            _all_conversations[_active_conv_id]["title"] = new_title[:60]
        return _history_html()

    # â”€â”€ Delete active conversation â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
    def delete_conv():
        global _active_conv_id
        if _active_conv_id in _all_conversations:
            del _all_conversations[_active_conv_id]
        _active_conv_id = ""
        clear_memory()
        return [], _history_html()

    # â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•
    # Layout
    # â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•â•
    with gr.Blocks(title="ðŸ¤– TaskTune Chatbot", css=CSS) as demo:

        gr.HTML("""
        <div style=\'text-align:center;padding:24px;
             background:linear-gradient(135deg,#0f0c29,#302b63,#24243e);
             border-radius:16px;margin-bottom:20px;
             border:1px solid #4f46e5;
             box-shadow:0 0 40px rgba(79,70,229,0.3);\'>
            <div style=\'font-size:44px;margin-bottom:8px;\'>ðŸ¤–</div>
            <h1 style=\'color:#c4b5fd;margin:0;font-size:26px;font-weight:700;letter-spacing:-0.3px;\'>
                TaskTune Chatbot
            </h1>
            <p style=\'color:#818cf8;margin:8px 0 0;font-size:14px;\'>
                Router: ðŸ’¬ chitchat &nbsp;|&nbsp; â“ clarify &nbsp;|&nbsp; ðŸ“š rag &nbsp;|&nbsp; ðŸ—„ï¸ sql &nbsp;|&nbsp; ðŸŒ web &nbsp;|&nbsp; ðŸ›Ÿ fallback
            </p>
        </div>
        """)

        with gr.Row():
            # â”€â”€ LEFT: History sidebar â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
            with gr.Column(scale=1, min_width=200):
                gr.Markdown("### ðŸ’¬ Conversations")
                new_btn     = gr.Button("ï¼‹ New chat", variant="primary")
                history_box = gr.HTML(value="<p style=\'color:#6b7280;font-size:13px;\'>No history yet.</p>")
                rename_txt  = gr.Textbox(label="Rename active chat", placeholder="New titleâ€¦", lines=1)
                rename_btn  = gr.Button("âœï¸ Rename", variant="secondary")
                del_btn     = gr.Button("ðŸ—‘ï¸ Delete chat", variant="secondary")
                # hidden helper for JS onclick load
                load_conv_id  = gr.Textbox(visible=False, elem_id="load_conv_id")
                load_conv_btn = gr.Button(visible=False, elem_id="load_conv_btn")

            # â”€â”€ MIDDLE: RAG settings â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
            with gr.Column(scale=1):
                gr.Markdown("### ðŸ“š Knowledge base (RAG)")
                file_upload = gr.File(
                    file_count="multiple",
                    file_types=[
                        ".pdf", ".docx", ".doc", ".xlsx", ".xls",
                        ".csv", ".txt", ".png", ".jpg", ".jpeg", ".webp",
                    ],
                )
                urls_input   = gr.Textbox(label="Websites (one URL per line)", lines=2)
                text_input   = gr.Textbox(label="Manual text", lines=2)
                build_btn    = gr.Button("Build knowledge base", variant="primary")
                build_status = gr.Textbox(label="Status", interactive=False)
                build_btn.click(
                    process_and_build,
                    [file_upload, urls_input, text_input],
                    [build_status],
                )
                gr.Markdown("### ðŸ—„ï¸ SQL / RBAC")
                role_dd = gr.Dropdown(
                    choices=["Admin", "Manager", "Employee"],
                    value="Admin",
                    label="SQL role (RBAC)",
                )
                emp_name = gr.Textbox(
                    label="Employee full name (only for Employee role)",
                    placeholder="e.g. Sara Mohamed",
                )

            # â”€â”€ RIGHT: Chat â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
            with gr.Column(scale=2):
                chatbot = gr.Chatbot(height=480, type="messages", allow_tags=False)
                msg  = gr.Textbox(placeholder="Ask anything â€” router picks RAG / SQL / Webâ€¦")
                with gr.Row():
                    send = gr.Button("Send", variant="primary")
                    clr  = gr.Button("Clear chat", variant="secondary")

        history_state = gr.State([])

        # â”€â”€ Wire events â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
        def do_send(m, h, r, n):
            msg_out, hist_out, hist_html = respond(m, h, r, n)
            return msg_out, hist_out, hist_html

        msg.submit(do_send, [msg, history_state, role_dd, emp_name],
                   [msg, history_state, history_box])
        send.click(do_send, [msg, history_state, role_dd, emp_name],
                   [msg, history_state, history_box])
        history_state.change(lambda h: h, inputs=[history_state], outputs=[chatbot])
        clr.click(lambda: ([], _history_html()), outputs=[history_state, history_box])

        new_btn.click(new_conv, outputs=[history_state, history_box])
        history_state.change(lambda h: h, inputs=[history_state], outputs=[chatbot])

        rename_btn.click(rename_conv, inputs=[rename_txt], outputs=[history_box])
        del_btn.click(delete_conv, outputs=[history_state, history_box])
        load_conv_btn.click(load_conv, inputs=[load_conv_id],
                            outputs=[history_state, history_box])
        history_state.change(lambda h: h, inputs=[history_state], outputs=[chatbot])

    demo.launch(share=share, show_error=True)


if __name__ == "__main__":
    launch_gradio(share=True)
````
